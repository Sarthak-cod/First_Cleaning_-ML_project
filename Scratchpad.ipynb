```python

import pandas as pd 

df = pd.read_csv("books.csv", on_bad_lines='skip')
df.head()
```

```python
# this shows us how many missing values  are hidden in each column
df.isnull().sum()
```

```python
df.dtypes
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect('books.db')
df.to_sql('books', conn, if_exists='replace', index=False)

print("DataFrame has been written to the SQLite database successfully.")
```

```python
# extract the top 5 books written in english using raw sql query
query = """SELECT title, authors, average_rating, publication_date
           FROM books
           WHERE language_code = 'eng'
           ORDER BY average_rating DESC
           LIMIT 5;
           """
# pull data directly from the SQL server into a fresh variable using pandas read_sql_query method
sql_extracted_df = pd.read_sql_query(query, conn)
sql_extracted_df
```

```python
# this parses the string data into a standardized DD-MM-YYYY format and stores it in a new column called 'parsed_publication_date'
# error= coerce  turn any completely broken entries  into a safe missing stamp value (NaT) instead of throwing an error

df['parsed_publication_date'] = pd.to_datetime(df['publication_date'], errors='coerce')

# verify the column type changed to datetime64

print("updated Column  Data Type  : ")
print(df['parsed_publication_date'].dtypes)
```
